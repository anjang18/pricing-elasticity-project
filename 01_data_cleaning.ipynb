## Final Analysis Sample

The analysis sample excludes observations without positive sales or positive
recorded prices, observations flagged as questionable by the dataset's `OK`
indicator, and UPCs identified from product metadata as non-cereal merchandise.

Extreme prices and quantities are not removed solely based on percentile
thresholds. Unusual observations are retained unless product metadata provides
evidence that they are not relevant cereal products.

```python
non_cereal_upcs = [
    317,          # Tony the Tiger T-shirt
    3828125053,   # Dominick's T-size shirt
    3828125057,   # Dominick's T-size shirt
    4300099100    # Holiday Home magazine shipper
]

analysis = sales[
    (sales["MOVE"] > 0) &
    (sales["UNIT_PRICE"] > 0) &
    (sales["OK"] == 1) &
    (~sales["UPC"].isin(non_cereal_upcs))
].copy()

print("Final analysis observations:", len(analysis))
print("Unique products:", analysis["UPC"].nunique())
print("Unique stores:", analysis["STORE"].nunique())
print("Weeks:", analysis["WEEK"].min(), "to", analysis["WEEK"].max())
```

```python
analysis[
    ["MOVE", "QTY", "PRICE", "UNIT_PRICE", "REVENUE", "PROFIT"]
].describe()
```

```python
print("Missing values:")
print(analysis.isna().sum())

print("\nDuplicate store-UPC-week observations:")
print(
    analysis.duplicated(
        subset=["STORE", "UPC", "WEEK"]
    ).sum()
)
```

```python
cereal_clean = sales[
    ~sales["UPC"].isin(non_cereal_upcs)
].copy()
```

```python
cereal_clean.to_parquet(
    "../data/processed/cereal_clean.parquet",
    index=False
)

analysis.to_parquet(
    "../data/processed/cereal_analysis.parquet",
    index=False
)

print("Files saved successfully.")
```
